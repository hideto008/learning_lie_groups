# 位相の公理の意味を考える

集合$X$上の任意の部分集合の集まりに対して、以下の位相の3公理を満たす位相を、集合$X$上の位相$\mathcal{T}$と呼ぶ。

1. 集合$X$と空集合$\text{\o}$が含まれる
2. 任意個の開集合の和集合が $\mathcal{T}$ に含まれる
3. 有限個の開集合の共通部分が $\mathcal{T}$ に含まれる

この公理2の「任意個の開集合の和集合」と、公理3の「有限個の開集合の共通部分」の意味について、実数の開集合を用いて捉えることを試みる。

### 1. 実数空間の開集合の定義

まず実数空間の開集合の定義を確認する。

実数空間 $\mathbb{R}$ において、集合 $\boldsymbol{U}$ が開集合であるとは、集合 $\boldsymbol{U}$ のすべての要素 $x$ において、その$x$ を中心とする十分小さな開区間がすべて$U$に含まれる、ということである。

$$
\boxed{
\forall \ x \in \boldsymbol{U}, \quad \exists \varepsilon > 0 , \quad (x - \varepsilon , x + \varepsilon) \subseteq \boldsymbol{U}
}
$$

例えば以下の3パターンの集合を考える。

- $(0,2)$

    これは 0 と 2 を含まない範囲である。ここで $x=0.001$ と限りなく0に近い値をとしても、$\varepsilon=0.0001$ のようにすると、$x - \varepsilon = 0.0009$ となり 0以下にはならない。どんなに小さな $x >0$としてもしてもそれよりも小さな$\varepsilon$は存在するため、$x - \varepsilon > 0$ とすることが出来る。同様に $x$ が 2に限りなく近い値でも $x + \varepsilon < 2$ とすることができる。

    よって、開集合の定義を満たしている。

- $[0,2]$

    これは 0 と 2 を含む範囲である。ここで $x=0$ とすると、$\varepsilon = 1e^{-16} >0$ のように極小の値としても $x - \varepsilon = - 1e^{-16} < 0 $ となって 集合の範囲を超えてしまう。これは $x=2$ として $x + \varepsilon > 2$ となるため、同じである。

    これは閉集合という。閉集合は、その補集合が開集合となる集合と定義される。

    今回の場合、補集合は 0未満の集合と2より大きい集合である。

    $$
    (-\infty, 0) \cup (2, \infty)
    $$

- $[0,2)$

    これは0を含み、2を含まない範囲である。上記で見たように、"2"側では開集合のように考えられるが、"x=0"では開集合ではない。よって、この集合は開集合での閉集合でもない。



### 2.通常の位相について

実数空間 $\mathbb{R}$ に通常の位相 $\mathcal{T}_{\text{usual}}$ があるとする。

ここで通常の位相とは、冒頭の集合における位相ではなく、実数空間 $\mathbb{R}$ を使って定義した開集合をすべて集めた位相のことである。

実数の開集合の定義は1章で見たように次のものであった。

$$
\boxed{
\forall \ x \in \boldsymbol{U}, \quad \exists \varepsilon > 0 , \quad (x - \varepsilon , x + \varepsilon) \subseteq \boldsymbol{U}
}
$$

例えば、以下はいずれも開集合となる。

$$
U_1 = (0,2), \quad U_2=(-1, 3) , \quad U_3=(-4, -2)  
$$

これより、実数空間 $\mathbb{R}$ の通常の位相の定義は以下となる。

$$
\boxed{
\mathcal{T}_{\text{usual}} = \{ \boldsymbol{U} \subseteq \mathbb{R} \ | \ \forall \ x \in \boldsymbol{U}, \ \exists \varepsilon > 0 , \ (x - \varepsilon , x + \varepsilon) \subseteq \boldsymbol{U}
\}
}
$$

これは集合 $U$は実数空間の$\mathbb{R}$の部分集合であり、かつ$U$は開集合であるというものである。したがって、

$$
(0,2) \in \mathcal{T}_{\text{usual}} , \quad [0,2] \notin  \mathcal{T}_{\text{usual}}
$$

$\mathcal{T}_{\text{usual}}$ は 実数空間 $\mathbb{R}$ 上の通常の位相である。よって、$(\mathbb{R}, \mathcal{T}_{\text{usual}})$ は通常の位相を持つ実数空間であるといえる。


### 3. 公理2 任意個の開集合の和集合

公理2が要求する「任意個」という性質について具体的に考えていく。

通常の位相に含まれる二つの開集合を考える。

$$
U_1 = (0,2) , \quad U_2 = (1,3)
$$

これらの和集合は以下となる。

$$
U_1 \cup U_2 = (0,3)
$$

$(0,3)$ は開集合であり、これは通常の位相に含まれる。よって公理2の「任意個の開集合の和集合が位相に含まれる」を満たしている。

ただしこれだけでは、任意個の和集合とはならない。任意個とは、どんな集合の集まりでもという意味であり、これは無限個の開集合でも成立するかを考える必要がある。

例えば、次の無限個の開集合を考える。

$$
U_n = (0, 1 - \frac{1}{n}) , \quad n = 2,3,4, ... , \quad n \in \mathbb{N}
$$

$n$には上限がないため、この和集合 $U$ は以下と定義できる。

$$
U = \bigcup_{n=2}^{\infty} U_n
$$

$U_n$を具体的に書くと、次のようになっている。つまり、$n$が大きくなるにつれて、区間 $U_n$ の右端 $1 - 1/n$ は1に近づいていく。

$$
\begin{aligned}
U_2 &= (0, 0.5) \\
U_3 &= (0, 0.6666...) \\
U_4 &= (0, 0.75) \\
& \vdots \\
U_{10} &= (0, 0.9) \\
& \vdots \\
U_{100} &= (0, 0.99) \\
& \vdots \\
\end{aligned}
$$

この関係により、それぞれの集合には次の包含関係がある。

$$
U_2 \subseteq U_3 \subseteq U_4 \subseteq \cdots \subseteq U_{10} \subseteq \cdots \subseteq U_{100}\cdots
$$ 

よって、$n$が有限個の$N$ならその集合和は$U_N$であり、

$$
U_2 \cup U_3 \cup U_4 \cup \cdots \cup U_N = U_N
$$

$U_N$は集合の集合である集合族$\{U_n | n \ge 2 \}$ の要素であり、開集合である。そして、$\mathcal{T}_{\text{usual}}$ に属している。

$$
U_N \in \mathcal{T}_{\text{usual}}
$$

しかし、和集合は以下のように無限個の集合の和である。

$$
U = \bigcup_{n=2}^{\infty} U_n
$$

これを式で書くと、

$$
\boxed{
x \in U \iff \exists n \ge 2 , \quad x \in U_n
}
$$

意味は、「$n \ge 2$ であるどれか一つの$U_n$に含まれる実数$x$をすべて集めた集合が$U$」である。

- (1) $ 0 < x < 1$ の場合
    
    $$
    x < 1 - \frac{1}{n}
    $$

    を満たす自然数$n$が存在すれば、$x \in U$ といえる。式を変形すると、

    $$
    n > \frac{1}{1-x}
    $$

    となる。例えば、x=0.999なら、

    $$
    n > \frac{1}{1- 0.999} = \frac{1}{0.001} = 1000
    $$

    を満たす$n=1001$を選べばよい。そして、どんな $0 < x < 1$ についても十分に大きな自然数$n$が存在する。したがって、$0 < x < 1$ はすべて和集合 $U$ に含まれる。

    $$
    (0,1) \subseteq U
    $$


- (2) $x=0$ または、$x \ge 1$ の場合

    $U_n$ の定義を確認すると、以下である。

    $$
    U_n = (0, 1 - \frac{1}{n}) , \quad n = 2,3,4, ... , \quad n \in \mathbb{N}
    $$

    よって、$U_n$に0は含まれず、(1)よりどんなに大きな$n$を指定しても1以上は $U_n$ に含まれない。また、$x \le 0 $ も同様に $U_n$ に含まれない。よって、和集合 $U$ は

    $$
    U \subseteq (0,1)
    $$

(1)と(2)より、集合の同一性として$(0,1) \subseteq U$ と $U \subseteq (0,1)$ が示せたので、集合和$U$は

$$
\boxed{
U = \bigcup_{n=2}^{\infty} U_n = (0,1)
}
$$

となる。

さて、ここで考えるべきば、有限個の$n$に対する $U_n$ に、和集合$U$は含まれるのか、というものである。

$$
U \subseteq U_n
$$

これを成立させるには、ある一つの$n$を固定して、その$U_n$が$U$のすべての点を含む必要がある。つまり、以下の条件となる。

$$
\exists n \ge 2 , \quad \forall x \in (0,1) , \quad x \in U_n
$$ 

しかしこれは成立しない。例えば $n = 1000$と固定すると、

$$
U_{1000} = (0, 1 - \frac{1}{1000}) = (0, 0.999)
$$

となる。この集合には、$x=0.9999$が含まれない。どんなに大きな有限の$n$を選んでも

$$
1 - \frac{1}{n} < 1
$$

なので、その$U_n$に含まれない $x \in (0,1)$が存在する。したがって、和集合$U$は$U_n$には含まれない。

$$
\boxed{
U \nsubseteq U_n \quad (\forall n \ge 2)
}
$$

以上より、各 $U_n$ は開集合であり、通常の位相 $\mathcal T_{\mathrm{usual}}$ に含まれる。

$$
U_n\in\mathcal T_{\mathrm{usual}}
$$

一方、無限和集合

$$
U=\bigcup_{n=2}^{\infty}U_n=(0,1)
$$

は、どの有限の $n$ に対する $U_n$ にも含まれない。

$$
U\nsubseteq U_n\qquad(\forall n\ge2)
$$

しかし、和集合 $U=(0,1)$ 自体が開集合の定義を満たすため、

$$
U\in\mathcal T_{\mathrm{usual}}
$$

が成立する。

つまり、各 $U_n$ が開集合であるだけでなく、それらの無限和集合 $U$ も通常の位相に含まれることが確認できた。

[このNote](./manifold.ipynb)の2章の有限集合 $X={a,b,c}$上の位相では、異なる開集合を無限個選ぶ例を考えることが出来なかった。そこで実数空間$\mathbb{R}$の通常の位相を用いて、無限個の異なる開集合の和集合も位相に含まれることを具体的に確認した。これは公理2「任意個の開集合の和集合も位相に含まれる」が成立する一例である。

### 4. 有限個の開集合の共通部分

まず2個の実数空間の開集合で確認する。例えば以下とする。

$$
U_1 = (0,3), \quad U_2 =(2,5)
$$

この共通部分は

$$
U_1 \cap U_2 = (2,3)
$$

であり、$(2,3)$は開集合なので、これは通常の位相$\mathcal{T}_{\text{usual}}$に含まれる。

$$
U_1 \cap U_2 =(2,3) \in \mathcal{T}_{\text{usual}}
$$

次に、無限個の場合の共通部分を考える。

次の無限個の開集合を考える。

$$
U_n = \left( -\frac{1}{n}, \frac{1}{n} \right) , \quad n = 1,2,3, ... \quad n \in \mathbb{N}
$$

具体的には

$$
\begin{aligned}
U_1 &= (-1, 1) \\
U_2 &= (-0.5, 0.5) \\
U_3 &= (-0.333..., 0.333...) \\
U_{10} &= (-0.1, 0.1) \\
U_{100} &= (-0.01, 0.01) \\
& \vdots
\end{aligned}
$$

となっており、$n$が増えるにつれてその範囲は0に近づいていく。

$n$が有限個の場合、例えば$U_1$から$U_{N}$までの共通部分は以下となる。

$$
\bigcap_{n=1}^{N} U_n = \left( - \frac{1}{N}, \frac{1}{N} \right)
$$

これは開集合である。

しかし、無限個の共通部分は次のようになる。

$$
U = \bigcap_{n=1}^{\infty} U_n = \{ 0 \}
$$

これは次のように確認することが出来る。

共通部分の定義は、全ての$U_n$に含まれる実数$x$を集めたものが$U$である。

$$
x \in U \Leftrightarrow \forall n \in \mathbb{N}, \quad x \in U_n
$$


- (1) $x=0$ の場合
    どんな自然数 $n \ge 1$ でも

    $$
    - \frac{1}{n} < 0 < \frac{1}{n}
    $$

    が成立する。これより、

    $$
    0 \in U_n \quad (\forall n \ge 1)
    $$

    であり、0はすべての$U_n$に含まれる。よって

    $$
    \{ 0 \} \subseteq U
    $$

- (2) $x>0$ の場合

    例えば$x=0.01$を考えると、

    $$
    U_{10} = (-0.1, 0.1)
    $$

    には含まれるが、

    $$
    U_{1000} = (-0.001, 0.001)
    $$

    には含まれない。つまり以下である。

    $$
    0.01 \notin U_{1000}
    $$

    共通部分に含まれるためには、全ての$U_n$に含まれる必要があるので、一つでも含まれない$U_n$があれば、その$x$は共通部分に含まれない。

    一般に $x>0$なら、

    $$
    n > \frac{1}{x}
    $$

    を満たす自然数 $n$ を選べば

    $$
    \frac{1}{n} < x
    $$

    となり、以下の関係となる。

    $$
    x \notin U_n
    $$

    これは、xをどんな小さな値を選んで、$n > 1/x$ を成立させる$n$が存在し、かつその場合は、$U_n$の右端について

    $$
    \frac{1}{n} < x
    $$

    とすることが出来る。よって、どんな $x>0$ に対しても、その$x$を含まない$U_n$が少なくとも一つ存在する。よって、$x>0$は共通部分 $U$ に含まれない。

    $$
    \forall x > 0, \quad \exists n \ge 1 , \quad x \notin U_n
    $$

- (3) $x<0$ の場合

    これも同様に、例えば $x=-0.01$ は

    $$
    U_{1000} = (-0.001, 0.001)
    $$

    には含まれない。

    一般に

    $$
    n > \frac{1}{|x|}
    $$

    を満たす自然数$n$を選べば

    $$
    |x| > \frac{1}{n}
    $$

    となるので、

    $$
    x \notin U_n
    $$

    となる。共通部分であるためには$x$は$U_n$に含まれる必要があるが、そうではないので、$x < 0 $ は $U$にも含まれない。

以上の(1),(2),(3)より、全ての$U_n$に含まれる実数は0だけなので、

$$
\boxed{
U = \bigcap_{n=1}^{\infty} U_n = \{ 0 \}
}
$$

この共通部分 $\{ 0 \}$は実数空間の開集合ではない。実際に$x=0$に対して、どんな $\varepsilon >0 $ を選んでも、$\{ 0 \}$は開集合の定義を満たせない。

$$
(- \varepsilon +0 , 0 + \varepsilon ) \nsubseteq \{0 \}
$$

よって、無限個の共通部分は通常の位相に含まれず、有限個のみが含まれる。

$$
\boxed{
\bigcap_{n=1}^{\infty} U_n = \{ 0 \} \notin \mathcal{T}_{\text{usual}}
}
$$

これより、有限個の開集合の共通部分は必ず開集合になるが、無限個の場合は開集合になるとは限らないということを示した。これより、公理3の「**有限個**の開集合の共通部分が位相に含まれる」という一つの具体例を確認することができた。


以上により、実数空間の通常の位相では、開集合の和集合は無限個であっても開集合となる。一方、開集合の共通部分は有限個なら開集合となるが、無限個の場合には開集合とならない例が存在する。

したがって、位相の公理2では「任意個」の和集合を要求し、公理3では「有限個」の共通部分のみを要求する理由を、具体例から確認できた。
