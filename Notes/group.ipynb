# 群(Group)について

## 定義

集合 $G$ と、その集合上の1つの演算 $\circ$ をセットと考える。

$$
\left(G, \ \circ \right)
$$

これが群であるためには、次の4つの条件を満たす必要がある。

1. 閉性

$$
a,\ b \ \in G \ \Rarr \ a \circ b \ \in G
$$

2. 結合則

$$
(a \circ b) \circ c = a \circ ( b \circ c)
$$

3. 単位元が存在

$$
\exist \ e \ \in G, \quad e \circ a = a \circ e = a
$$

4. 逆元が存在

$$
\forall \ a \ \in G, \quad \exist \ a^{-1} \ \in G , \quad a^{-1} \circ a = a \circ a^{-1} = e
$$

上記のように、集合 $G$ に対して一つの演算 $\circ$ があり、それらに "1. 閉性、2. 結合則、3. 単位元が存在、4. 逆元が存在" があれば、群となる。　

## 簡単な例

### 整数の加法

$$
\left(\mathbf{Z}, + \right)
$$

$a, b \in \mathbf{Z}$ とすると、

$$
a + b \ \in \mathbf{Z}
$$

となり、整数と整数を足しても有理数や実数にはならず整数のままなので、"閉性"がある。

$$
a + (b + c) = (a + b ) + c
$$ 

であるため、"結合則"を満たしている。

$$
a + 0 = 0 + a = a
$$

であるので、"単位元 $0$ が存在"している。

$$
a + (-a) = (-a) + a = 0
$$

であるので、"逆元が存在"している。

そして、

$$
a +b = b + a
$$

のようになるので、$\left(\mathbf{Z}, + \right)$ は可換群 (Abel 群) でもある。



## 行列の群

次の回転行列の集合 $SO(3)$ を考える。

$$
SO(3) = \left\{R \in \mathbf{R^{3x3}} | R^T \ R = I , \ \det R = 1 \right\}
$$

演算として、行列積を考える。

$$
\left( SO(3) , \times \right)
$$

閉性は $R_1 , R_2 \ \in SO(3)$ として、$R_1 R_2 \ \in SO(3)$ であると示せればよい。
