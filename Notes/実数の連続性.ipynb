# 実数の連続性についてのメモ

$\epsilon - \delta$ 論法による実数の連続性についての理解を記述する。

## 1. 定義

関数$f(x)$が連続であるとは、以下の定義を満たすことをいう。

$$
\forall \varepsilon > 0 , \quad \exists \delta >0 , \forall x : |x - a| < \delta \Rightarrow |f(x) - f(a)| < \varepsilon
$$

この定義を読み解くと、

- ある$a$について考える。
- 関数 $f$ の出力である $f(a)$ と、任意の $x$ についての出力 $f(x)$ の差($|f(x) - f(a)|$)として $\varepsilon$ を指定するとき、
- それを保証できる入力範囲$|x - a|$ を指定できる$\delta$が存在する。

ということである。

例えば$f(x)=2x$のような正の比例関係であれば、$\varepsilon$ を大きく指定すると、$x$ の範囲はそれに伴い広くとることができる。そして、広くなった$x$ により入力の範囲$|x - a|$は大きくなるが、その範囲を指定する $\delta$ についても大きくなる。というような流れがある。

## 2. 具体例

### 2.1 $f(x) = 3x$

まず、$a=2$として考える。

すると、出力範囲は

$$
|f(x) - f(a)| = |3x - 6| < \varepsilon
$$

となる。ここから入力の範囲$|x - a|$は次のようになる。

$$
3|x - 2| < \varepsilon \quad \Rightarrow  |x - 2| < \frac{\varepsilon}{3} = \delta
$$

ここから $\varepsilon > 0$ をどのように変化させても $\delta$ は $\varepsilon$ を3 で割れば求まるので、$\delta$ は存在することが出来る。

よって、この関数は$a=2$において連続であるといえる。

$a$を指定せずに考えると、出力範囲は

$$
|f(x) - f(a)| = |3x - 3a| < \varepsilon
$$

入力範囲は

$$
3|x - a| < \varepsilon \Rightarrow |x - a| < \delta = \frac{\varepsilon}{3}
$$

であるので、この関数はいかなる$a$に対しても連続であるといえる。

### 2.2 不連続関数の例

$$
f(x) = \begin{cases}
0 &\quad x < 0 \\
1 &\quad x \ge 0 
\end{cases}
$$

ここで$a=0$を考える。

出力側は以下である。

$$
|f(x) - f(a)| = |f(x) - 1| < \varepsilon
$$

$x$が正の場合、$f(x)=1$より$|f(x) - 1|=0$となり、$x$が負の場合、$f(x)=0$より$|f(x) - 1|=1$となる。$\forall \varepsilon$ であるが、$\varepsilon > 1$である必要がある。

そして、$\varepsilon = 1.1$を指定した場合は$x$の値はどんな値でも取りえるので、入力範囲である $|x - a| = |x|$ はどんな値でも取りえるということになる。しかし、この場合は、$x$が何か決まれば$|x| < \delta$ となる$\delta$を指定すればよい。

#### 不連続となる理由

関数が$a=0$において連続であるためには、任意の$\varepsilon$に対して、条件を満たす$\delta>0$が存在する必要がある。

ここで、$\varepsilon = 0.5$ を指定した場合、任意の$\delta >0$に対して、

$$
x = -\frac{\delta}{2}
$$

という入力を選ぶとする。すると、入力側の条件は、以下のように満たすことが出来る。

$$
|x - a| = \left| -\frac{\delta}{2} \right| =  \frac{\delta}{2} < \delta
$$

しかし、$x<0$であるため、

$$
f(x) = 0
$$

これより、

$$
|f(x) - f(0)| = |0 - 1| = 1 \ \cancel{<} \ \varepsilon = 0.5
$$

となって、最初の$\varepsilon$の設定が適切ではなかったということになる。よって、任意の$\varepsilon$に対して条件を満たす$\delta$が存在しないということになり、この関数は$a=0$において不連続であるといえる。



今見たように、不連続性の証明には、

$$
\exists \varepsilon > 0 , \quad \forall \delta >0 , \exists x : |x - a| < \delta \land  |f(x) - f(a)| \ge \varepsilon
$$

を示す必要がある。

今回の場合は $\varepsilon = 0.5$ を固定して、任意の$\delta >0$ に対して、

$$
x = - \frac{\delta}{2}
$$

と選べば、必ず入力側の条件を満たし、出力側の条件を満たさなくなる。つなり$\delta$をどれだけ小さくしても、条件に反する$x$が存在するということになる。
