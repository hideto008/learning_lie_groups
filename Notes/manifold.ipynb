# 多様体(Maniforld)についての概略

ここではLie群の理解で必要な多様体についての概略と簡単な例を述べる。

多様体は「全体としては曲がっていたり複雑な形をしていても、十分に小さいでは普通のユークリッド空間のように扱える空間」というものである。

この ***十分に小さい範囲ではユークリッド空間にように扱える*** とは、その小さい範囲とユークリッド空間の開集合の間に、連続的な逆写像を持つ1体1の座標変換が存在するということである。

この定義の言葉を一つづつ確認する。

## 1. 開集合について

まず、開集合の定義は以下である。

### 1.1 開集合の定義

実数空間 $\mathbb{R}$ において、集合 $\boldsymbol{U}$ が開集合であるとは、集合 $\boldsymbol{U}$ のすべての要素 $x$ において、その$x$ の微小変化させた範囲 $(x - \varepsilon , x + \varepsilon)$ も 集合 $\boldsymbol{U}$ となる $\varepsilon$ が存在する、ということである。

$$
\forall \ x \in \boldsymbol{U}, \quad \exists \varepsilon > 0 , \quad (x - \varepsilon , x + \varepsilon) \subset \boldsymbol{U}
$$

例えば以下の3パターンの集合を考える。

- $(0,2)$

    これは 0 と 2 を含まない範囲である。ここで $x=0.001$ と限りなく0に近い値をとしても、$\varepsilon=0.0001$ のようにすると、$x - \varepsilon = 0.00009$ となり 0以下にはならない。これは $x$ が 2に限りなく近い値でも同じである。

    よって、開集合の定義を満たしている。

- $[0,2]$

    これは 0 と 2 を含む範囲である。ここで $x=0$ とすると、$\varepsilon = 1e^{-16} >0$ のように極小の値としても $x - \varepsilon = - 1e^{-16} < 0 $ となって 集合の範囲を超えてしまう。これは $x=2$ として同じである。

    これは閉集合という。閉集合は、その補集合が開集合となる集合と定義される。

    今回の場合、補集合は 0未満の集合と2より大きい集合である。

    $$
    (-\infty, 0) \cup (2, \infty)
    $$



## 2. 多様体の具体例

単位円と単位球を用いて多様体とは何かを考える。

### 2.1 単位円

単位円は全体的に曲がった形をしている。

直交座標系$(x,y)$は、角度$\theta$を用いて次のように表せる。

$$
\begin{bmatrix}
x \\ y 
\end{bmatrix} =
\begin{bmatrix}
\cos(\theta) \\
\sin(\theta)
\end{bmatrix}
$$
