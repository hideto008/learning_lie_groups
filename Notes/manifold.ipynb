# 多様体(Maniforld)についての概略

ここではLie群の理解で必要な多様体についての概略と簡単な例を述べる。

多様体は「全体としては曲がっていたり複雑な形をしていても、十分に小さいでは普通のユークリッド空間のように扱える空間」というものである。

この ***十分に小さい範囲ではユークリッド空間にように扱える*** とは、その小さい範囲とユークリッド空間の開集合の間に、連続的な逆写像を持つ1体1の座標変換が存在するということである。

この定義の言葉を以下の順に理解する。



## 1. 集合・写像・逆像

- 集合はある要素の集まりである。数字だけでなく、コインの(裏、表)、何月何日といった状態なども要素として考えることが出来る。

- 写像というのは、元となる集合の各要素に対して、ある集合のいずれか１つの要素と対応させる規則である。

- 逆像とは写像により対応させた要素の、元の集合の要素ということである。

数式で書くと次のようになる。

集合$V$と$U$に属する要素を次のように記述する。

$$
V = \{a, b, c, d \}, \quad U =\{ h, i, j \}
$$

$V$から$U$への写像$f$は次のように記述する。このとき$V$は始域(しいき)、$U$は終域(しゅういき)と呼ばれる。

$$
f : V \rightarrow U
$$

写像は元の集合の各要素に対するものであるため、例えば次のようなものである。

$$
\begin{aligned}
f(a) = h, \quad f(b) = h, \quad f(c) = i, \quad f(d) = i
\end{aligned}
$$

このように写像による対応させた$U$の$\{h,i \}$を値域(ちいき)と呼ぶ。。値域は終域$U$の部分集合である。これを$A \sube U$ と記載する。

逆像は写像が定義されたうえで成り立つものである。今回は写像$f$の値域$A$に対して次のようになる。

$$
f^{-1}(A) = \{ x \in V | f(x) \in A \}
$$

そして、写像と対応させると、次のようになる。逆関数のような記述になっているが、実際の関数ではなく、値は何だったかということを示すものが逆像である。

$$
f^{-1}(\{h\}) = \{a,b\}, \quad f^{-1}(\{i\}) = \{c,d\}
$$

$j \in U$ の逆像は存在しないため、空集合$\text{\O}$となる。

$$
f^{-1}(\{j\}) = \text{\O}
$$

絵で描くと次のような形になる。

</p>

終域全体の逆像は、以下より始域となる。

$$
\begin{aligned}
f^{-1}(U) &= f^{-1}(\{h\}) \ \cup \ f^{-1}(\{i\}) \ \cup \ f^{-1}(\{j\}) \\
&= \{a,b\} \ \cup \ \{c,d \} \ \cup  \text{\O} \\
&= \{a,b, c,d \} = V
\end{aligned}
$$

また空集合$\text{\O}$の逆像も空集合である。

$$
f^{-1}(\text{\O}) = \text{\O}
$$



## 2. 位相空間と開集合

位相(Topology)とは、集合の部分集合のうち、どれを開集合として扱うかを定める数学的な構造である。

集合 $X$ に対して、開集合の集まり $\mathcal{T}$ を定めたものを位相空間と呼び、$(X, \mathcal{T})$ と表す。

$\mathcal{T}$ は任意の部分集合の集まりではなく、以下の位相の公理を満たす必要がある。

1. 集合$X$と空集合$\text{\O}$が含まれる
2. 任意個の開集合の和集合が $\mathcal{T}$ に含まれる
3. 有限個の開集合の共通部分が $\mathcal{T}$ に含まれる

この公理を満たす $\mathcal{T}$ の要素を開集合と呼ぶ。

ここで公理2の **任意個** とは、個数に制限を設けず、どのような開集合の組み合わせでも成立することを意味する。

また、公理3の**有限個** とは、有限後の開集合をどのように選んでも成立することを意味する。

天下り的だが、この公理をいったん飲み込み進める。

例えば、集合$X$を以下とする。

$$
X = \{ a, b, c\}
$$

#### $\mathcal{T}_1$

まず、$\mathcal{T}_1$ を以下として考える。

$$
\mathcal{T}_1 = \{ \text{\O} , X\}
$$

まず、これは公理1を満たしている。<br>

次に公理2の任意個の和集合について考える。その組み合わせを上げると、以下でありこれらは$\mathcal{T}_1$ に含まれる。よって、公理2を満たしている。

$$
\begin{aligned}
\text{\O} \cup \text{\O} = \text{\O} , \quad X \cup X = X , \quad \text{\O} \cup X = X
\end{aligned}
$$

最後に、公理3の有限個の共通部分について考える。これも組み合わせを上げると以下であり、これらは$\mathcal{T}_1$に含まれる。よって公理3を満たしている。

$$
\begin{aligned}
\text{\O} \cap \text{\O} = \text{\O} , \quad X \cap X = X , \quad \text{\O} \cap X = \text{\O}
\end{aligned}
$$

よって$\mathcal{T}_1$の各要素は開集合であり、$(X, \mathcal{T}_1)$ は位相空間である、となる。

#### $\mathcal{T}_2$

次に、以下を考える。

$$
\mathcal{T}_2 = \{ \text{\O}, {a}, X\}
$$

これは、公理1について満たしている。

公理2の和集合について、$\mathcal{T}_1$ との差分を示すと、以下であるため、いずれも$\mathcal{T}_2$の要素である。よって公理2を満たしている。

$$
\begin{aligned}
\text{\O} \cup  \{a\}= \{a\} , \quad \{a\} \cup \{a\} = \{a\} , \quad \{a\} \cup X = X , \quad \text{\O} \cup  \{a\} \cup X = X
\end{aligned}
$$

公理3の共有部分について、$\mathcal{T}_1$ との差分を示すと以下であるため、いずれも$\mathcal{T}_2$の要素である。よって公理3を満たしている。

$$
\begin{aligned}
\text{\O} \cap \{a\} = \text{\O} , \quad \{a\} \cap \{a\} = \{a\} , \quad \{a\} \cap X = \{a\} , \quad \text{\O} \cap \{a\} \cap X = \text{\O}
\end{aligned}
$$

よって$\mathcal{T}_2$の各要素は開集合であり、$(X, \mathcal{T}_2)$ は位相空間である、となる。

#### $\mathcal{T}_3$

次に、以下を考える。

$$
\mathcal{T}_3 = \{ \text{\O}, {a}, {b}, X\}
$$

これは、公理1について満たしている。

公理2の和集合について、次の組み合わせがあり、これは$\mathcal{T}_3$の要素ではない、よって公理2を満たしていない。

$$
\begin{array}{c}
\{a\} \cup  \{b\}= \{a, b\}
\end{array}
$$

公理3の共有部分について、$\mathcal{T}_2$との差分を示すと以下の組み合わせである。これらはすべて$\mathcal{T}_3$の要素であり、公理3を満たす。

$$
\begin{array}{c}
\{a\} \cap  \{b\} = \text{\O} , \quad \{b\} \cap  \{b\} = \{b\}, \\
\text{\O} \cap \{a\} \cap  \{b\} = \text{\O}, \quad X \cap \{a\} \cap  \{b\} = \text{\O}, \\
\text{\O} \cap \{a\} \cap  \{b\} \cap X = \text{\O}
\end{array}
$$

よって$\mathcal{T}_3$は公理2を満たさないので、$(X, \mathcal{T}_3)$ は位相空間ではない、となる。

ここまでは一般の集合$X$に対して、位相の公理から開集合を定義した。一方、実数空間$\mathbb{R}$では、距離を用いて開集合を定義することができる。このように定義された開集合の公理を満たすため、実数空間の通常の位相となる。

#### 公理の意味合い

先ほど天下り的に受け入れた公理を実数の開集合でとらえてもう少しわかりやすくとらえてみる。

### 2.1 実数空間の開集合の定義

まず実数空間の開集合の定義を確認する。

実数空間 $\mathbb{R}$ において、集合 $\boldsymbol{U}$ が開集合であるとは、集合 $\boldsymbol{U}$ のすべての要素 $x$ において、その$x$ の微小変化させた範囲 $(x - \varepsilon , x + \varepsilon)$ も 集合 $\boldsymbol{U}$ となる $\varepsilon$ が存在する、ということである。

$$
\forall \ x \in \boldsymbol{U}, \quad \exists \varepsilon > 0 , \quad (x - \varepsilon , x + \varepsilon) \subset \boldsymbol{U}
$$

例えば以下の3パターンの集合を考える。

- $(0,2)$

    これは 0 と 2 を含まない範囲である。ここで $x=0.001$ と限りなく0に近い値をとしても、$\varepsilon=0.0001$ のようにすると、$x - \varepsilon = 0.0009$ となり 0以下にはならない。どんなに小さな $x >0$としてもしてもそれよりも小さな$\varepsilon$は存在するため、$x - \varepsilon > 0$ とすることが出来る。同様に $x$ が 2に限りなく近い値でも $x + \varepsilon < 2$ とすることができる。

    よって、開集合の定義を満たしている。

- $[0,2]$

    これは 0 と 2 を含む範囲である。ここで $x=0$ とすると、$\varepsilon = 1e^{-16} >0$ のように極小の値としても $x - \varepsilon = - 1e^{-16} < 0 $ となって 集合の範囲を超えてしまう。これは $x=2$ として $x + \varepsilon > 2$ となるため、同じである。

    これは閉集合という。閉集合は、その補集合が開集合となる集合と定義される。

    今回の場合、補集合は 0未満の集合と2より大きい集合である。

    $$
    (-\infty, 0) \cup (2, \infty)
    $$

- $[0,2)$

    これは0を含み、2を含まない範囲である。上記で見たように、"2"側では開集合のように考えられるが、"x=0"では開集合ではない。よって、この集合は開集合での閉集合でもない。
    

```python
import matplotlib.pyplot as plt
import numpy as np

n = np.linspace(2, 10, 100)
u1 = -1 + 1 / n
u2 = +1 - 1 / n

plt.plot(n, u1, label='u1')
plt.plot(n, u2, label='u2')
plt.legend()
plt.legend()
plt.show()
```

## 2. 多様体の具体例

単位円と単位球を用いて多様体とは何かを考える。

### 2.1 単位円

単位円は全体的に曲がった形をしている。

直交座標系$(x,y)$は、角度$\theta$を用いて次のように表せる。

$$
\begin{bmatrix}
x \\ y 
\end{bmatrix} =
\begin{bmatrix}
\cos(\theta) \\
\sin(\theta)
\end{bmatrix}
$$
